## Data preparation
The dataset contains student, course, mentor, platform, payment, discount and target-revenue fields.

```python
df.info()
df.shape
```

## Q1 – Course category lookup and enrollment count

```python
course_category = {"C201": "Analytics", "C203": "Development"}
df["Category"] = df["CourseID"].map(course_category).fillna("Not Found")
category_counts = df["Category"].value_counts().rename_axis("Category").reset_index(name="EnrollmentCount")
category_counts
```

## Q2 – Total Mobile enrollment

```python
mobile_enrollment = (df["Platform"] == "Mobile").sum()
mobile_enrollment
```

## Q3 – Total payment amount for Data Analytics

```python
data_analytics_payment = df.loc[df["CourseName"] == "Data Analytics", "PaymentAmount"].sum()
data_analytics_payment
```

## Q4 – Lookup mentor names using Student ID

```python
lookup_ids = pd.DataFrame({"StudentID": ["S1206", "S1010", "S1011"]})
q4 = lookup_ids.merge(df[["StudentID", "MentorName"]], on="StudentID", how="left")
q4["MentorName"] = q4["MentorName"].fillna("NOT FOUND")
q4
```

## Q5 – Target achievement status

```python
df["Performance Status"] = np.where(df["PaymentAmount"] > df["TargetRevenue"], "Target Achieved", "Below Target")
target_achieved = (df["Performance Status"] == "Target Achieved").sum()
below_target = (df["Performance Status"] == "Below Target").sum()
target_achieved_pct = target_achieved / len(df) * 100
pd.DataFrame({"Metric": ["Target Achieved Count", "Below Target Count", "Target Achieved %"],
              "Value": [target_achieved, below_target, target_achieved_pct]})
```

## Q6 – Validate discount codes and discount percentage

```python
valid_codes = {"DISC10", "DISC20", "DISC30"}
df["Validated Discount %"] = np.where(
    df["DiscountCode"].isin(valid_codes) & df["DiscountCode"].notna(),
    df["DiscountPercent"].fillna(0),
    0
)
q6_summary = pd.DataFrame({
    "Metric": ["Valid discount-code rows", "Invalid/missing discount-code rows"],
    "Value": [int((df["Validated Discount %"] > 0).sum()), int((df["Validated Discount %"] == 0).sum())]
})
q6_summary
```

## Q7 – INDEX/MATCH style lookup of platform by student name

```python
names = ["Rohit Shah", "Riya Singh", "Sneha Gupta", "Sneha Iyer", "Karan Patil"]
# Equivalent pandas lookup: first matching record, as INDEX/MATCH returns the first match.
q7 = pd.DataFrame({"StudentName": names})
q7["Platform"] = q7["StudentName"].map(df.drop_duplicates("StudentName").set_index("StudentName")["Platform"])
q7
```

## Q8 – Revenue before and after discount

```python
df["Net Revenue"] = df["PaymentAmount"] * (1 - df["DiscountPercent"].fillna(0))
q8 = pd.DataFrame({
    "Metric": ["Total Revenue Before Discount", "Total Revenue After Discount"],
    "Value": [df["PaymentAmount"].sum(), df["Net Revenue"].sum()]
})
q8
```

## Q9 – Analytics category on Web platform

```python
analytics_web_count = ((df["Category"] == "Analytics") & (df["Platform"] == "Web")).sum()
q9 = pd.DataFrame({"Metric": ["Analytics rows", "Analytics + Web rows"],
                  "Value": [(df["Category"] == "Analytics").sum(), analytics_web_count]})
q9
```

## Q10 – XLOOKUP style return of student details

```python
q10_ids = ["S1000"]
q10 = df[df["StudentID"].isin(q10_ids)][["StudentID","StudentName","CourseID","CourseName","BatchID","MentorName","Platform","PaymentAmount","DiscountCode","DiscountPercent"]]
q10
```

## Q11 & Q12 – Category × Platform payment pivot

```python
payment_pivot = pd.pivot_table(
    df, index="Category", columns="Platform", values="PaymentAmount", aggfunc="sum", fill_value=0, margins=True, margins_name="Grand Total"
)
payment_pivot
```

```python
net_revenue_pivot = pd.pivot_table(
    df, index="Category", columns="Platform", values="Net Revenue", aggfunc="sum", fill_value=0, margins=True, margins_name="Grand Total"
)
net_revenue_pivot
```

## Q13 – Mentor/platform and monthly payment summaries

```python
mentor_platform_pivot = pd.pivot_table(
    df, index="MentorName", columns="Platform", values="PaymentAmount", aggfunc="sum", fill_value=0, margins=True, margins_name="Grand Total"
)
mentor_platform_pivot
```

```python
monthly_payment = df.assign(Month=df["PaymentDate"].dt.strftime("%b")).pivot_table(
    index="Month", values="PaymentAmount", aggfunc="sum"
).reindex(["Jan", "Feb", "Mar"])
monthly_payment
```

```python
target_achievement_ratio = df["PaymentAmount"].sum() / df["TargetRevenue"].sum() * 100
target_achievement_ratio
```
